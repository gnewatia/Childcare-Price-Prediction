```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
import warnings
import requests
import os
import janitor as j
warnings.filterwarnings('ignore')
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
data_raw = pd.read_csv('NDCP2022.csv', encoding='utf-8', encoding_errors='ignore')
```

```python
data_raw=(pd.DataFrame(data_raw)
    .clean_names()  # Clean column names
    .drop_constant_columns()  # Remove constant columns
)
```

```python
data_adjusted = (
    data_raw
    .remove_columns(['state_abbreviation','state_fips','county_fips_code'])
    .remove_columns([ 'mfi','mhi', 'me', 'fme', 'mme','mfccsa',
 'mcinfant',
 'mctoddler',
 'mcpreschool',
 'mfccinfant',
 'mfcctoddler',
 'mfccpreschool']) 
    .remove_columns([col for col in data_raw.columns if col.endswith((
        "bto5", "6to11", "12to17", "18to23", "24to29", 
        "30to35", "36to41", "42to47", "48to53", "54tosa"))])  # Drop columns ending with these patterns
    .remove_columns([col for col in data_raw.columns if col.endswith("_flag")]) 
      .remove_columns([col for col in data_raw.columns if col.endswith("_state")])
     # Drop columns ending with "_flag"  # Drop columns starting with "x75"
)

data_adjusted=(data_adjusted.remove_columns([col for col in data_adjusted.columns if col.startswith("_75")])
               .remove_columns([col for col in data_adjusted.columns if col.startswith("i")]))
```

```python
# number of rows and columns of the dataframe
print('Shape of the data is: ',data_adjusted.shape)
```

```python
data_final=data_adjusted[data_adjusted['studyyear']==2022]
data_final.to_csv('data_final.csv',index=False)
```
